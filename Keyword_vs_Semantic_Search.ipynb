# Similarity Search vs Keyword Search
## Introduction

Traditional keyword-based search retrieves results only when the exact query words
are present in the document.

However, human language is flexible — the same meaning can be expressed using
different words, synonyms, or sentence structures.

This notebook demonstrates why keyword search fails for semantic queries
and how similarity search using embeddings provides better results.
## Objective

The objective of this notebook is to compare:

- Traditional keyword-based search
- Modern semantic similarity search using sentence embeddings

and demonstrate why similarity search performs better for natural language queries.
## What This Notebook Contains

This notebook includes:

- An explanation of keyword-based search
- A simple implementation of keyword search
- Limitations of keyword search
- Introduction to semantic similarity search
- Use of sentence embeddings
- Cosine similarity for measuring relevance
- A comparison between keyword and similarity search


#PART 1: KEYWORD SEARCH
## Why Do We Need Better Search?

Modern applications such as:
- Chatbots
- Search engines
- Recommendation systems
- Retrieval-Augmented Generation (RAG)

require understanding **meaning**, not just matching words.

This motivates the need for semantic similarity search.
## Keyword Search

Keyword search:
- Looks for exact word matches
- Does not understand meaning
- Works well for structured or exact queries
- Fails for synonyms and paraphrased queries


```python
# Given documents
documents = [
    "Melbourne is a vibrant city in southeastern Australia, known for its rich culture and modern lifestyle",
    "It is famous for its coffee culture, street art, and diverse food scene.",
    "Melbourne is often ranked among the world’s most livable cities."
]

query = "best city to live"  # user query

results = []

for doc in documents:
    if query.lower() in doc.lower():
        results.append(doc)

print("Keyword Search Results:")
print(results)
```

## Limitations of Keyword Search

- Cannot detect synonyms
- Cannot handle paraphrasing
- Order-sensitive
- No understanding of context

As a result, no documents are returned even though relevant information exists.

##PART-2 Similarity Search

Similarity search works by:
1. Converting text into numerical vectors (embeddings)
2. Capturing semantic meaning
3. Comparing vectors using similarity metrics

This allows the system to understand intent rather than exact words.
## What Are Embeddings?

Embeddings are numerical representations of text where:
- Sentences with similar meanings are closer in vector space
- Different meanings are farther apart

Pre-trained transformer models are used to generate embeddings.


```python
!pip install sentence-transformers numpy

import numpy as np
from sentence_transformers import SentenceTransformer
```

```python
model = SentenceTransformer("all-MiniLM-L6-v2")
```

## Data Preparation

The same documents are split into chunks and converted into embeddings.


```python
chunks = [
    "Melbourne is a vibrant city in southeastern Australia, known for its rich culture and modern lifestyle",
    "It is famous for its coffee culture, street art, and diverse food scene.",
    "Melbourne is often ranked among the world’s most livable cities."
]
```

## Generating Text Embeddings


```python
chunk_embeddings = model.encode(chunks)
```

```python
chunk_embeddings.shape
```

## Query Embedding


```python
query = "Why is Melbourne considered one of the world’s most livable cities?"
query_embedding = model.encode(query)
```

##Cosine Similarity Function

Cosine similarity measures how close two vectors are in semantic space.


```python
def cosine_similarity(vec1, vec2):
    return np.dot(vec1, vec2) / (np.linalg.norm(vec1) * np.linalg.norm(vec2))
```

## Computing Similarity Scores


```python
for i, chunk in enumerate(chunks):
    score = cosine_similarity(query_embedding, chunk_embeddings[i])
    print(f"Chunk {i+1} similarity score: {score:.4f}")
```

##  Retrieving Top Matches


```python
scores = [
    cosine_similarity(query_embedding, emb)
    for emb in chunk_embeddings
]

top_k = sorted(
    zip(chunks, scores),
    key=lambda x: x[1],
    reverse=True
)

for text, score in top_k:
    print(f"Score: {score:.4f} | Text: {text}")
```

## Keyword Search vs Similarity Search

| Feature | Keyword Search | Similarity Search |
|------|--------------|----------------|
| Understands meaning | No | Yes |
| Handles paraphrasing | No | Yes |
| Uses embeddings | No | Yes |
| Suitable for RAG | No | Yes |


## Observations

- Keyword search fails when exact words are not present
- Similarity search retrieves relevant content despite paraphrasing
- Embeddings capture semantic meaning effectively
- Cosine similarity provides meaningful ranking
